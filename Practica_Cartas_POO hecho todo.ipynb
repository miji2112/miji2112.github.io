**Ejercicio 4**: Escribir un método que nos permita comparar cartas por igualdad

```python
class Carta:
    def __init__(self, numero, palo):
        self.numero = numero
        self.palo = palo

    def __eq__(self, other):
        # Si tienen el mismo número y el mismo palo, son iguales (True)
        # Si no, son diferentes (False)
        return self.numero == other.numero and self.palo == other.palo
       
# Creamos dos cartas para probar
carta_a = Carta(1, "Espada")
carta_b = Carta(1, "Espada")
carta_c = Carta(1, "Oro")

# Ejemplo 1: Comparando cartas idénticas
if carta_a == carta_b:
    print("¡Empate! Son la misma carta.")
else:
    print("Son cartas distintas.")

# Ejemplo 2: Mismo número pero distinto palo
if carta_a == carta_c:
    print("¡Empate!")
else:
    print("Son distintas (tienen el mismo número, pero cambia el palo).")
```

**Ejercicio 5**: Escribir una clase mazo, que construya el mazo de cartas españolas. Escribir un método que devuelve cuantas cartas hay en el mazo.

```python
class Carta:
    def __init__(self, numero, palo):
        self.numero = numero
        self.palo = palo

class Mazo:
    def __init__(self):
        self.cartas = []
        palos = ["Espada", "Basto", "Oro", "Copa"]
        # El mazo español tradicional tiene del 1 al 12 (sin 8 y 9)
        numeros = [1, 2, 3, 4, 5, 6, 7, 10, 11, 12]
        
        # Creamos todas las combinaciones de cartas
        for palo in palos:
            for numero in numeros:
                nueva_carta = Carta(numero, palo)
                self.cartas.append(nueva_carta)

    def cuantas_cartas(self):
        # len() nos dice cuántos elementos hay en la lista
        return len(self.cartas)

# --- Ejemplo de uso ---
mi_mazo = Mazo()
print("Cantidad de cartas en el mazo:", mi_mazo.cuantas_cartas())
```
